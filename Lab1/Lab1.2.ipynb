## 1. Încărcarea Overlay-ului de bază

Placa PYNQ-Z2 include un Overlay standard numit **Base Overlay**. Acesta configurează logica programabilă și oferă acces la mai multe resurse ale plăcii, inclusiv LED-uri, switch-uri și butoane.

Încărcarea Overlay-ului se face din Python:


````python
from pynq.overlays.base import BaseOverlay

base = BaseOverlay("base.bit")
````

Prin această instrucțiune, fișierul `base.bit` este încărcat în **Programmable Logic (PL)**.

Conceptual, fluxul este:

```text
Python / Jupyter
      │
      ▼
BaseOverlay("base.bit")
      │
      ▼
Configurarea FPGA-ului
      │
      ▼
Programmable Logic (PL)
```


## 2. Accesarea LED-urilor

LED-urile utilizator sunt disponibile prin obiectul `base.leds`.

Verificați numărul de LED-uri disponibile:


````python
print("Număr LED-uri:", len(base.leds))
````

Selectăm primul LED al plăcii:


````python
led0 = base.leds[0]
````

Aprindeți LED0:


````python
led0.on()
````

Verificați fizic placa. LED0 trebuie să fie aprins.

Stingeți LED0:


````python
led0.off()
````

Metoda `toggle()` inversează starea curentă a LED-ului:


````python
led0.toggle()
````

## 3. Generarea unei secvențe hardware

Folosim Python pentru a comanda repetat LED-ul:


````python
import time

for i in range(10):
    led0.toggle()
    time.sleep(0.2)

led0.off()
````

## 4. Controlul tuturor LED-urilor

Putem accesa cele patru LED-uri prin index:


````python
leds = base.leds

for i, led in enumerate(leds):
    print(f"LED{i}: {led}")
````

Realizați o secvență în care LED-urile se aprind succesiv de la LED0 la LED3:


````python
for led in leds:
    led.off()

for led in leds:
    led.on()
    time.sleep(0.3)
    led.off()
````
