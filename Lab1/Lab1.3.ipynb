## 1. Încărcarea Base Overlay


````python
from pynq.overlays.base import BaseOverlay
import time

base = BaseOverlay("base.bit")

print("Switch-uri:", len(base.switches))
print("LED-uri    :", len(base.leds))
````

## 2. Citirea switch-urilor

PYNQ-Z2 are două switch-uri utilizator: **SW0** și **SW1**. Starea fiecăruia poate fi citită cu metoda `read()`.


````python
sw0 = base.switches[0].read()
sw1 = base.switches[1].read()

print("SW0 =", sw0)
print("SW1 =", sw1)
````

Cele două switch-uri pot reprezenta o valoare binară pe 2 biți:

```text
SW1 SW0
 0   0  -> 0
 0   1  -> 1
 1   0  -> 2
 1   1  -> 3
```


````python
value = base.switches[0].read() | (base.switches[1].read() << 1)
print(f"Valoare = {value} ({value:02b})")
````

## 3. Copierea stării switch-urilor pe LED-uri

În exemplul următor, SW0 controlează LED0, iar SW1 controlează LED1. Rulați celula și modificați poziția switch-urilor.

Opriți execuția cu **Kernel → Interrupt**.


````python
try:
    while True:
        for i, sw in enumerate(base.switches):
            if sw.read():
                base.leds[i].on()
            else:
                base.leds[i].off()
        time.sleep(0.05)
except KeyboardInterrupt:
    pass
finally:
    for led in base.leds:
        led.off()
````
