# Lab1.4 — Citirea butoanelor

În acest task vom citi cele patru butoane utilizator ale plăcii PYNQ-Z2 și vom observa diferența dintre o **stare logică** și un **eveniment de apăsare**.


## Obiective

- citirea celor patru butoane;
- reprezentarea stării lor ca valoare pe 4 biți;
- controlul LED-urilor cu butoanele;
- observarea comportamentului unui buton menținut apăsat.


## 1. Inițializare


```python
from pynq.overlays.base import BaseOverlay
import time

base = BaseOverlay("base.bit")

print("Butoane :", len(base.buttons))
print("LED-uri :", len(base.leds))
```

## 2. Citirea unui buton

Apăsați BTN0 și executați celula. Repetați apoi fără a apăsa butonul.


```python
print("BTN0 =", base.buttons[0].read())
```

## 3. Citirea tuturor butoanelor

Starea celor patru butoane poate fi împachetată într-o valoare pe 4 biți.


```python
value = 0
for i, button in enumerate(base.buttons):
    value |= (button.read() & 1) << i

print(f"Buttons = {value:04b}")
```

Rulați celula următoare și apăsați diferite butoane, inclusiv două simultan. Se afișează doar schimbările de stare.

Opriți execuția cu **Kernel → Interrupt**.


```python
previous = None

try:
    while True:
        current = 0
        for i, button in enumerate(base.buttons):
            current |= (button.read() & 1) << i

        if current != previous:
            print(f"Buttons = {current:04b}")
            previous = current

        time.sleep(0.02)
except KeyboardInterrupt:
    pass
```

## 4. Afișarea stării butoanelor pe LED-uri

Cât timp un buton este apăsat, LED-ul corespunzător trebuie să fie aprins.


```python
try:
    while True:
        for i, button in enumerate(base.buttons):
            if button.read():
                base.leds[i].on()
            else:
                base.leds[i].off()
        time.sleep(0.02)
except KeyboardInterrupt:
    pass
finally:
    for led in base.leds:
        led.off()
```

## 5. Observație

Metoda `read()` returnează **starea curentă** a butonului. Dacă un buton rămâne apăsat, programul îl va citi ca `1` la fiecare iterație a buclei.

În taskul următor vom transforma această stare într-un **eveniment unic de apăsare**, necesar pentru implementarea unui contor.
