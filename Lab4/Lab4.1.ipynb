## 1. Pregătirea imaginii de test

Folosim aceeași imagine în toate notebook-urile Lab4.


````python
from pathlib import Path

IMAGE_PATH = Path("images/fruits.jpg")

if not IMAGE_PATH.exists():
    raise FileNotFoundError(
        "Nu a fost găsită imaginea Lab4/images/fruits.jpg. "
        "Verificați că repository-ul a fost copiat complet pe placă."
    )

print("Image:", IMAGE_PATH)
````

## 2. Încărcarea imaginii


````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_bgr = cv2.imread(str(IMAGE_PATH), cv2.IMREAD_COLOR)

if image_bgr is None:
    raise RuntimeError("Imaginea nu a putut fi încărcată.")

print("OpenCV version:", cv2.__version__)
print("Python type    :", type(image_bgr))
print("NumPy dtype   :", image_bgr.dtype)
print("Shape         :", image_bgr.shape)
````

Pentru o imagine color, `shape` are forma:

```text
(height, width, channels)
```

În mod obișnuit, fiecare canal este reprezentat prin `uint8`, adică **8 biți = 1 byte per canal**.


````python
height, width, channels = image_bgr.shape

print("Width    :", width)
print("Height   :", height)
print("Channels :", channels)
print("Pixels   :", width * height)
print("Bytes    :", image_bgr.nbytes)
print("MiB      :", image_bgr.nbytes / (1024**2))
````

## 3. BGR vs RGB

OpenCV încarcă implicit imaginile color în ordinea **BGR**, în timp ce Matplotlib așteaptă **RGB**.


````python
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(image_rgb)
plt.title("Original image")
plt.axis("off")
plt.show()
````

## 4. Accesarea unui pixel


````python
y = height // 2
x = width // 2

pixel_bgr = image_bgr[y, x]
print(f"Pixel ({x}, {y}) BGR =", pixel_bgr)
print("B =", int(pixel_bgr[0]))
print("G =", int(pixel_bgr[1]))
print("R =", int(pixel_bgr[2]))
````

## 5. Separarea canalelor


````python
b, g, r = cv2.split(image_bgr)

print("B shape:", b.shape)
print("G shape:", g.shape)
print("R shape:", r.shape)
print("One channel bytes:", r.nbytes)
````

## 6. Relația rezoluție–memorie

Pentru o imagine RGB/BGR `uint8`:

```text
memorie = width × height × 3 bytes
```


````python
for w, h in ((320, 240), (640, 480), (1280, 720), (1920, 1080)):
    nbytes = w * h * 3
    print(f"{w:4d}x{h:<4d} -> {nbytes:9d} bytes -> {nbytes/(1024**2):6.2f} MiB")
````
